# SpeechSep on Colab

Runs the full separation → diarization → transcription pipeline on a Colab GPU.
The package is installed from the GitHub repo, so nothing is duplicated here.

**Before running:** make sure the latest code is pushed to
`https://github.com/Constantine1824/speechd.git` (branch `fix/sample-rate-and-cleanups`),
and set the runtime to GPU: *Runtime → Change runtime type → T4 GPU*.

```python
# Install the package from GitHub (re-run after every push to pick up changes)
%pip install -q "git+https://github.com/Constantine1824/speechd.git@main"

import torch
print("CUDA available:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "none — enable GPU runtime!")
```

```python
# Upload an audio file (or skip this cell and set AUDIO_PATH to a Drive path instead)
from google.colab import files
uploaded = files.upload()
AUDIO_PATH = next(iter(uploaded))
print("Using:", AUDIO_PATH)
```

```python
from speechsep.main import run
from speechsep.schemas import PipelineConfig

config = PipelineConfig(
    num_speakers=2,          # 2 or 3 speakers; for single speaker use denoise_only=True
    # denoise_only=True,
    auto_num_speakers=False, # True = estimate speaker count during clustering
    whisper_model="small",   # tiny / base / small / medium / large-v3
    device="cuda",
    compute_type="float16",  # float16 pairs best with cuda
    language=None,           # None = auto-detect, or e.g. "en"
    save_path="transcript.json",
    save_fmt="json",         # json / plain / rttm / srt
    print_output=True,
)

results = run(AUDIO_PATH, config=config)
```
